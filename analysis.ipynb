**Research question:** To what extent are Pokémon card rarity and age associated with their current market value?

This notebook loads the **cleaned** collection snapshot. Collection and cleaning live in `collect.py` and `clean.py`.

## Setup

```python
from pathlib import Path
import pandas as pd

DATA = Path("data")
clean_path = DATA / "processed" / "cards_clean.csv"
sample_path = DATA / "sample" / "cards_clean_sample.csv"

df = pd.read_csv(clean_path, parse_dates=["release_date"])
sample = pd.read_csv(sample_path, parse_dates=["release_date"])
print(f"Clean rows: {len(df):,} across {df['set_id'].nunique()} sets")
print(f"Sample rows: {len(sample):,}")
df.head()
```

## Cleaning output (summary)

Rows require rarity, set release date (for age), and a positive TCGPlayer market price. See `data/data_dictionary.md`.

```python
summary = pd.DataFrame({
    "metric": ["n_cards", "n_sets", "price_min_usd", "price_median_usd", "price_max_usd", "age_min_years", "age_max_years"],
    "value": [
        len(df),
        df["set_id"].nunique(),
        df["market_price_usd"].min(),
        df["market_price_usd"].median(),
        df["market_price_usd"].max(),
        df["age_years"].min(),
        df["age_years"].max(),
    ],
})
display(summary)
display(df["rarity"].value_counts().to_frame("n"))
display(sample.head(10))
```
